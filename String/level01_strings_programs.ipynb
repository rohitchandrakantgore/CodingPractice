```python
# reverese a string with all types of methods
str = 'RohitGore'
#slicing
res_slicing = str[::-1]
print(res_slicing)

# Without built-in slicing (loop / two-pointer approach)
res_manual = ""
for char in str:
    res_manual = char + res_manual

print(res_manual)
```

```python
# check palindrome 
str= "madam"
# Built-in / Slicing
is_palindrome_slicing = str==str[::-1]
print(is_palindrome_slicing)

str="sir"
# Without built-in (Two-pointer technique)
is_palindrome_manual = True
left,right = 0, len(str)-1
while left < right:
    if str[left]!= str[right]:
        is_palindrome_manual = False
        break
    left+=1
    right-=1
print(is_palindrome_manual)
```

```python
#Count the number of vowels and consonants in a string.

str = 'abracadabra'
vowels_set = set("aeiouAEIOU")

# Built-in (sum with generator)
v_count_bi = sum(1 for c in str if c in vowels_set)
c_count_bi = sum(1 for c in str if c.isalpha() and c not in vowels_set)

print(v_count_bi)
print(c_count_bi)

#manual loop
def get_count(str):
    vowles = 0
    consonants= 0
    for char in str:
        if char in vowels_set:
            vowles+=1
        else:
            consonants+=1
    return vowles, consonants

print(get_count(str))

```

```python
# Find the frequency of every character in a string.
str = 'abracadabra'
frequeny = {}

# Built-in (collections.Counter)
from collections import Counter
freq_builtin = dict(Counter(str))
print(freq_builtin)

#method 2
def check_frequency(str):
    for i in range(len(str)):
        if str[i] not in frequeny.keys():
            frequeny[str[i]]=1
        else:
            frequeny[str[i]]=frequeny[str[i]]+1
    return frequeny

#method 3
def check_frequency_2(str):
    result={}
    for char in str:
        result[char] = result.get(char,0)+1
    return result


print(check_frequency(str))
print(check_frequency_2(str))
```

```python
#Remove all spaces from a string
str = "ArithmeticError Exception"

#method 1
without_space = "".join(str.split(" "))
print(without_space)

# method 2
no_space= ""
for char in str:
    if char != " ":
        no_space +=char
print(no_space)

#method 3
replaced = str.replace(" ","")
print(replaced)
```

```python
#Convert lowercase characters to uppercase
str = "AssertionError Exception"

#method 1
upper_str = str.upper()
print(upper_str)

#method 2
upper = ""
for char in str:
    if "a"<= char <="z":
        upper += chr(ord(char) - 32)
    else:
        upper += char
print(upper)
```

```python
#Convert uppercase characters to lowercase and lowercase characters to uppercase.
str = "AssertionError Exception"

#method 1
swap_str = str.swapcase()
print(swap_str)

#method 2
swapped = ""
for char in str:
    if "a"<= char <="z":
        swapped += chr(ord(char) - 32)
    elif "A"<= char <= "Z":
        swapped += chr(ord(char) + 32)
    else:
        swapped += char
print(swapped)
```
