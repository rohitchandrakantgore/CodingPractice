```python
#Sum of List Elements
list_numbers= [1000000001, 1000000002, 1000000003, 1000000004, 1000000005]
def sum_of_list(lst):
    total = 0
    for n in lst:
        total += n
    return total

result = sum_of_list(list_numbers)
print(result)
```

```python
#Sorting of List Elements with sort() method
list_numbers= [0, 0, 0, 2, 2, 2, 1, 1, 1]
def sort_of_list(lst):
    lst.sort()
    print("Sorted List:", *lst)
    return lst

result = sort_of_list(list_numbers)
print(result)
```

```python
#Sorting of List Elements with sorted() method
list_numbers= [0, 0, 0, 2, 2, 2, 1, 1, 1]
def sort_of_list(lst):
    sorted_lst = sorted(lst)
    print("Sorted List:", *sorted_lst)
    return sorted_lst

result = sort_of_list(list_numbers)
print(result)
```

```python
#sorting without using sort() and sorted() method
list_numbers= [0, 0, 0, 2, 2, 2, 1, 1, 1]
def sort_of_list(lst):
    for i in range(len(lst)):
        for j in range(i + 1, len(lst)):
            if lst[i] > lst[j]:
                lst[i], lst[j] = lst[j], lst[i]
    print("Sorted List:", *lst)
    return lst

result = sort_of_list(list_numbers)
print(result)
```

```python
#find kth largest element in a list without using sort() and sorted() method
list_numbers= [0, 0, 0, 2, 2, 2, 1, 1, 1]
k = 3
def kth_largest(lst, k):
    for i in range(len(lst)):
        for j in range(i + 1, len(lst)):
            if lst[i] < lst[j]:
                lst[i], lst[j] = lst[j], lst[i]
    print("Sorted List:", *lst)
    return lst[k - 1]

print(kth_largest(list_numbers, k))
```

```python
#largest possible number from a list of non-negative integers without using sort() and sorted() method
list_numbers= [3, 30, 34, 5, 9]
def largest_number(lst):
    for i in range(len(lst)):
        for j in range(i + 1, len(lst)):
            if str(lst[i]) + str(lst[j]) < str(lst[j]) + str(lst[i]):
                lst[i], lst[j] = lst[j], lst[i]
    print("Sorted List:", *lst)
    return ''.join(map(str, lst))

result = largest_number(list_numbers)
print(result)
```

```python
#minimum number of operations to reach number from 0 
# where we have only two operations allowed: multiply by 2 and add 1
# with operation printed

def min_operations(n):
    operations = 0
    while n > 0:
        if n % 2 == 0:
            n //= 2
            print("Operation:", n * 2)
        else:
            n -= 1
            print("Operation:", n + 1)
        operations += 1
    return operations

print(min_operations(10))  # Output: 4
```
