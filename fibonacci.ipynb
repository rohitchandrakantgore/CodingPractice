```python
#fibonnaci series program
# fibonnaci series means the sum of the previous two numbers in the series. The first two numbers in the series are 0 and 1. 
# The next number is 0 + 1 = 1, then 1 + 1 = 2, then 1 + 2 = 3, and so on.

def fibonacci_series(n):
    a, b = 0, 1
    series = []
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series

print("Fibonacci series up to n terms:")
print(fibonacci_series(10))
```

```python
#sum of the fibonacci series program with recursion
def sum_of_fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return sum_of_fibonacci(n - 1) + sum_of_fibonacci(n - 2)

print("Sum of Fibonacci series up to n terms:")
n_terms = 10
print(sum_of_fibonacci(n_terms))
```
